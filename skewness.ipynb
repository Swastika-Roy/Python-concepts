```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Normal Distribution (Symmetric)
np.random.seed(42)
normal_data = np.random.normal(loc=0, scale=1, size=1000)
skew_normal = stats.skew(normal_data)

# Plot 1: Histogram with KDE
axes[0].hist(normal_data, bins=30, density=True, alpha=0.6, color='skyblue', edgecolor='black')
sns.kdeplot(normal_data, ax=axes[0], color='darkblue', linewidth=2)
axes[0].axvline(np.mean(normal_data), color='red', linestyle='--', label=f'Mean: {np.mean(normal_data):.2f}')
axes[0].axvline(np.median(normal_data), color='green', linestyle=':', label=f'Median: {np.median(normal_data):.2f}')
axes[0].set_title(f'Symmetric Distribution\\nSkewness: {skew_normal:.3f}')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Box Plot
axes[1].boxplot(normal_data, vert=False, patch_artist=True,
                boxprops=dict(facecolor='lightblue'))
axes[1].set_title('Box Plot (Symmetric)')
axes[1].set_xlabel('Value')
axes[1].grid(True, alpha=0.3)

# Q-Q Plot
stats.probplot(normal_data, dist="norm", plot=axes[2])
axes[2].set_title('Q-Q Plot (Normal)')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Symmetric Distribution Statistics:")
print(f"  Mean: {np.mean(normal_data):.4f}")
print(f"  Median: {np.median(normal_data):.4f}")
print(f"  Mode: Not well-defined for continuous data")
print(f"  Skewness: {skew_normal:.4f}")
print("  Relationship: Mean ≈ Median ≈ Mode")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Exponential Distribution (Right Skew)
exp_data = np.random.exponential(scale=2, size=1000)
skew_exp = stats.skew(exp_data)

# Plot 1: Histogram with KDE
axes[0].hist(exp_data, bins=30, density=True, alpha=0.6, color='salmon', edgecolor='black')
sns.kdeplot(exp_data, ax=axes[0], color='darkred', linewidth=2)
axes[0].axvline(np.mean(exp_data), color='red', linestyle='--', label=f'Mean: {np.mean(exp_data):.2f}')
axes[0].axvline(np.median(exp_data), color='green', linestyle=':', label=f'Median: {np.median(exp_data):.2f}')
axes[0].set_title(f'Right (Positive) Skew\\nSkewness: {skew_exp:.3f}')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Box Plot
axes[1].boxplot(exp_data, vert=False, patch_artist=True,
                boxprops=dict(facecolor='lightcoral'))
axes[1].set_title('Box Plot (Right Skew)')
axes[1].set_xlabel('Value')
axes[1].grid(True, alpha=0.3)

# Violin Plot
axes[2].violinplot(exp_data, vert=False, showmeans=True)
axes[2].set_title('Violin Plot')
axes[2].set_xlabel('Value')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Right Skew Distribution Statistics:")
print(f"  Mean: {np.mean(exp_data):.4f}")
print(f"  Median: {np.median(exp_data):.4f}")
print(f"  Mode: ~0 (theoretical for exponential distribution)")
print(f"  Skewness: {skew_exp:.4f}")
print("  Relationship: Mode < Median < Mean")
print("  Tail extends to the RIGHT")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Beta Distribution (Left Skew)
beta_data = np.random.beta(a=5, b=2, size=1000)
skew_beta = stats.skew(beta_data)

# Plot 1: Histogram with KDE
axes[0].hist(beta_data, bins=30, density=True, alpha=0.6, color='lightgreen', edgecolor='black')
sns.kdeplot(beta_data, ax=axes[0], color='darkgreen', linewidth=2)
axes[0].axvline(np.mean(beta_data), color='red', linestyle='--', label=f'Mean: {np.mean(beta_data):.2f}')
axes[0].axvline(np.median(beta_data), color='green', linestyle=':', label=f'Median: {np.median(beta_data):.2f}')
axes[0].set_title(f'Left (Negative) Skew\\nSkewness: {skew_beta:.3f}')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Box Plot
axes[1].boxplot(beta_data, vert=False, patch_artist=True,
                boxprops=dict(facecolor='lightgreen'))
axes[1].set_title('Box Plot (Left Skew)')
axes[1].set_xlabel('Value')
axes[1].grid(True, alpha=0.3)

# ECDF Plot
from statsmodels.distributions.empirical_distribution import ECDF
ecdf = ECDF(beta_data)
axes[2].plot(ecdf.x, ecdf.y, color='darkgreen', linewidth=2)
axes[2].set_title('ECDF Plot (Left Skew)')
axes[2].set_xlabel('Value')
axes[2].set_ylabel('Cumulative Probability')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Left Skew Distribution Statistics:")
print(f"  Mean: {np.mean(beta_data):.4f}")
print(f"  Median: {np.median(beta_data):.4f}")
print(f"  Mode: {(5 - 1) / (5 + 2 - 2):.4f} (theoretical)")
print(f"  Skewness: {skew_beta:.4f}")
print("  Relationship: Mean < Median < Mode")
print("  Tail extends to the LEFT")
```

Step 3: Variance
Population Variance (σ²):
Sum of (x - μ)² = 16+4+0+4+16 = 40
σ² = 40/5 = 8
Sample Variance (s²):s² = 40/(5-1) = 40/4 = 10
### Step 4: Standard Deviation

**Population SD (σ):**
σ = √8 = 2.8284
Sample SD (s):s = √10 = 3.1623

### Step 5: Skewness

**Population Skewness:**
Sum of (x - μ)³ = -64-8+0+8+64 = 0
γ₁ = [Sum(x - μ)³ / n] / σ³
   = [0/5] / (2.8284³)
   = 0

**Sample Skewness (Fisher-Pearson coefficient):**
g₁ = [n/((n-1)(n-2))] × [Sum(x - μ)³] / s³
   = [5/(4×3)] × [0] / (3.1623³)
   = 0

**Population Kurtosis:**
Sum of (x - μ)⁴ = 256+16+0+16+256 = 544
β₂ = [Sum(x - μ)⁴ / n] / σ⁴
   = [544/5] / (2.8284⁴)
   = 108.8 / 64 = 1.7
