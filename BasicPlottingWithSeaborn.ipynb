```python
#import neccessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')
```

```python
#seaborn comes with built-in datasets for practice
print("Available datasets")
datasets = sns.get_dataset_names()
print(datasets)
```

```python
#load sample datasets
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
titanic = sns.load_dataset('titanic')
diamonds = sns.load_dataset('diamonds')
flights = sns.load_dataset('flights')
```

**Basic  PLOT TYPES**

**Line plot**

```python
#create figure with subslots
fig,axes = plt.subplots(1,2,figsize=(15,5))

#basic line plot
sns.lineplot(data=flights,x='year',y='passengers',ax=axes[0])
axes[0].set_title('Basic Line plot',fontsize=14,fontweight='bold')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('NO of Passengers')

#LINE PLOT WITH HUE(Multiple lines)
sns.lineplot(data=flights,x='year',y='passengers',hue='month',ax=axes[1],palette='tab10')

axes[1].set_title('Line plot with Multiple categories',fontsize=14,fontweight='bold')
axes[1].set_xlabel('Year')
axes[1].set_ylabel('NO of Passengers')
axes[1].legend(bbox_to_anchor=(1.05,1),loc="upper left")
plt.tight_layout()
plt.show()
```

**Scatter Plot**

```python
tips.head()
```

```python
fig,axes = plt.subplots(2,2,figsize=(15,12))

#basic scatter plot
sns.scatterplot(data=tips,x='total_bill',y='tip',ax=axes[0,0])
axes[0,0].set_title('Basic Scatter plot',fontsize=14,fontweight='bold')

#scatter plot with hue
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='time',ax=axes[0,1])
axes[0,1].set_title('Scatter plot with Hue',fontsize=14,fontweight='bold')

#scatter plot with size
sns.scatterplot(data=tips,x='total_bill',y='tip',size='size',ax=axes[1,0])
axes[1,0].set_title('Scatter plot with size',fontsize=14,fontweight='bold')


#scatter plot with hue,size and style
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='day',size='size',style='time',ax=axes[1,1])
axes[1,1].set_title("Scatter plot  with hue,size and style",fontsize=14,fontweight='bold')

plt.tight_layout()
plt.show()
```

 **bar plot**

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Basic bar plot
sns.barplot(data=tips, x='day', y='total_bill', ax=axes[0, 0])
axes[0, 0].set_title('Basic Bar Plot', fontsize=14, fontweight='bold')
axes[0, 0].set_ylabel('Average Total Bill')

# Bar plot with hue
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', ax=axes[0, 1])
axes[0, 1].set_title('Bar Plot with Hue', fontsize=14, fontweight='bold')
axes[0, 1].set_ylabel('Average Total Bill')

# Horizontal bar plot
sns.barplot(data=tips, x='total_bill', y='day', ax=axes[1, 0], orient='h')
axes[1, 0].set_title('Horizontal Bar Plot', fontsize=14, fontweight='bold')
axes[1, 0].set_xlabel('Average Total Bill')

# Bar plot with custom estimator
sns.barplot(data=tips, x='day', y='total_bill',
            estimator=np.median, ax=axes[1, 1])
axes[1, 1].set_title('Bar Plot with Median', fontsize=14, fontweight='bold')
axes[1, 1].set_ylabel('Median Total Bill')

plt.tight_layout()
plt.show()
```

**Count plot**

```python
fig,axes = plt.subplots(2,2,figsize=(15,12))

#basic count plot
sns.countplot(data=tips,x='day',ax=axes[0,0])
axes[0,0].set_title(' Count Plot - Days',fontsize=14,fontweight='bold')
axes[0,0].set_ylabel('Count')

#count plot with hue
sns.countplot(data=tips,x='day',hue='sex',ax=axes[0,1])
axes[0,1].set_title('Count Plot  with Hue',fontsize=14,fontweight='bold')
axes[0,1].set_ylabel('Count')


#horizontal count plot
sns.countplot(data=tips,y='day',ax=axes[1,0])
axes[1,0].set_title('Horizontal Count Plot',fontsize=14,fontweight='bold')
axes[1,0].set_xlabel('Count')


#count plot with custom order and palette
order = ['Thur','Fri','Sat','Sun']
sns.countplot(data=tips,x='day',order=order,palette='Set2',ax=axes[1,1])
axes[1,1].set_title('Count Plot with Custom Order',fontsize=14,fontweight='bold')
axes[1,1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

**categorial plots**

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Basic box plot
sns.boxplot(data=tips, x='day', y='total_bill', ax=axes[0, 0])
axes[0, 0].set_title('Basic Box Plot', fontsize=14, fontweight='bold')

# Box plot with hue
sns.boxplot(data=tips, x='day', y='total_bill', hue='smoker', ax=axes[0, 1])
axes[0, 1].set_title('Box Plot with Hue', fontsize=14, fontweight='bold')

# Horizontal box plot
sns.boxplot(data=tips, x='total_bill', y='day', ax=axes[1, 0], orient='h')
axes[1, 0].set_title('Horizontal Box Plot', fontsize=14, fontweight='bold')

# Box plot with custom width and palette
sns.boxplot(data=tips, x='day', y='total_bill',
            width=0.5, palette='Set3', ax=axes[1, 1])
axes[1, 1].set_title('Box Plot with Custom Style', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

**VIOLIN PLOT**

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Basic violin plot
sns.violinplot(data=tips, x='day', y='total_bill', ax=axes[0, 0])
axes[0, 0].set_title('Basic Violin Plot', fontsize=14, fontweight='bold')

# Violin plot with hue
sns.violinplot(data=tips, x='day', y='total_bill',
               hue='sex', ax=axes[0, 1])
axes[0, 1].set_title('Violin Plot with Hue', fontsize=14, fontweight='bold')

# Split violin plot
sns.violinplot(data=tips, x='day', y='total_bill',
               hue='sex', split=True, ax=axes[1, 0])
axes[1, 0].set_title('Split Violin Plot', fontsize=14, fontweight='bold')

# Violin plot with inner quartiles
sns.violinplot(data=tips, x='day', y='total_bill',
               inner='quartile', palette='muted', ax=axes[1, 1])
axes[1, 1].set_title('Violin Plot with Quartiles', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

**CAT PLOT**

```python
# catplot creates a figure-level plot
# It's more flexible than axes-level plots

# Box plot using catplot
g1 = sns.catplot(data=tips, x='day', y='total_bill',
                 kind='box', height=5, aspect=1.5)
g1.fig.suptitle('Catplot - Box', fontsize=14, fontweight='bold', y=1.02)
plt.show()

# Violin plot with hue using catplot
g2 = sns.catplot(data=tips, x='day', y='total_bill',
                 hue='sex', kind='violin', height=5, aspect=1.5)
g2.fig.suptitle('Catplot - Violin with Hue', fontsize=14, fontweight='bold', y=1.02)
plt.show()

# Bar plot with column faceting
g3 = sns.catplot(data=tips, x='day', y='total_bill',
                 col='time', kind='bar', height=5, aspect=0.8)
g3.fig.suptitle('Catplot - Bar with Facets', fontsize=14, fontweight='bold', y=1.02)
plt.show()
```

**DISTRIBUTION PLOT**

```python
fig,axes = plt.subplots(2,2,figsize=(15,12))

#basic histogram
sns.histplot(data=tips,x='total_bill',ax=axes[0,0])
axes[0,0].set_title('Histogram',fontsize=14,fontweight='bold')
axes[0,0].set_xlabel('Total Bill')


#histogram with KDE
sns.histplot(data=tips,x='total_bill',kde=True,ax=axes[0,1])
axes[0,1].set_title('Histogram with KDE',fontsize=14,fontweight='bold')


#histogram with hue
sns.histplot(data=tips,x='total_bill',hue='sex',ax=axes[1,0])
axes[1,0].set_title('Histogram with Hue',fontsize=14,fontweight='bold')


#histogram with multiple variables(stacked)
sns.histplot(data=tips,x='total_bill',hue='time',multiple='stack',ax=axes[1,1])
axes[1,1].set_title('Histogram with multiple variables',fontsize=14,fontweight='bold')
```

**KDE PLOT**

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Basic KDE plot
sns.kdeplot(data=tips, x='total_bill', ax=axes[0, 0])
axes[0, 0].set_title('Basic KDE Plot', fontsize=14, fontweight='bold')

# KDE plot with fill
sns.kdeplot(data=tips, x='total_bill', fill=True, ax=axes[0, 1])
axes[0, 1].set_title('KDE Plot with Fill', fontsize=14, fontweight='bold')

# KDE plot with hue
sns.kdeplot(data=tips, x='total_bill', hue='time',
            fill=True, ax=axes[1, 0])
axes[1, 0].set_title('KDE Plot with Hue', fontsize=14, fontweight='bold')

# Bivariate KDE plot
sns.kdeplot(data=tips, x='total_bill', y='tip',
            fill=True, ax=axes[1, 1])
axes[1, 1].set_title('Bivariate KDE Plot', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

**REGRESSION**

**SCATTER**

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Basic regression plot
sns.regplot(data=tips, x='total_bill', y='tip', ax=axes[0, 0])
axes[0, 0].set_title('Basic Regression Plot', fontsize=14, fontweight='bold')

# Regression plot without confidence interval
sns.regplot(data=tips, x='total_bill', y='tip',
            ci=None, ax=axes[0, 1])
axes[0, 1].set_title('Regression Plot without CI', fontsize=14, fontweight='bold')

# Polynomial regression (order=2)
sns.regplot(data=tips, x='total_bill', y='tip',
            order=2, ax=axes[1, 0])
axes[1, 0].set_title('Polynomial Regression (order=2)',
                     fontsize=14, fontweight='bold')

# Robust regression
sns.regplot(data=tips, x='total_bill', y='tip',
            robust=True, ax=axes[1, 1])
axes[1, 1].set_title('Robust Regression', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

**LM PLOT**

```python
# Basic lmplot
g1 = sns.lmplot(data=tips, x='total_bill', y='tip',
                height=5, aspect=1.5)
g1.fig.suptitle('Basic LM Plot', fontsize=14, fontweight='bold', y=1.02)
plt.show()

# LM plot with hue
g2 = sns.lmplot(data=tips, x='total_bill', y='tip',
                hue='smoker', height=5, aspect=1.5)
g2.fig.suptitle('LM Plot with Hue', fontsize=14, fontweight='bold', y=1.02)
plt.show()

# LM plot with column faceting
g3 = sns.lmplot(data=tips, x='total_bill', y='tip',
                col='time', height=5, aspect=0.8)
g3.fig.suptitle('LM Plot with Facets', fontsize=14, fontweight='bold', y=1.02)
plt.show()

# LM plot with row and column faceting
g4 = sns.lmplot(data=tips, x='total_bill', y='tip',
                col='time', row='smoker', height=4, aspect=1)
g4.fig.suptitle('LM Plot with Row and Column Facets',
                fontsize=14, fontweight='bold', y=1.01)
plt.show()
```

**RESIDUAL PLOT**

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Residual plot
sns.residplot(data=tips, x='total_bill', y='tip', ax=axes[0])
axes[0].set_title('Residual Plot', fontsize=14, fontweight='bold')
axes[0].axhline(y=0, color='red', linestyle='--', linewidth=2)

# Residual plot with lowess
sns.residplot(data=tips, x='total_bill', y='tip',
              lowess=True, ax=axes[1])
axes[1].set_title('Residual Plot with LOWESS', fontsize=14, fontweight='bold')
axes[1].axhline(y=0, color='red', linestyle='--', linewidth=2)

plt.tight_layout()
plt.show()
```

```python
#HEATMAP
# Create correlation matrix
tips_numeric = tips.select_dtypes(include=[np.number])
correlation = tips_numeric.corr()

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Basic heatmap
sns.heatmap(correlation, ax=axes[0])
axes[0].set_title('Basic Heatmap', fontsize=14, fontweight='bold')

# Heatmap with annotations
sns.heatmap(correlation, annot=True, fmt='.2f',
            cmap='coolwarm', center=0, ax=axes[1])
axes[1].set_title('Annotated Heatmap', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
#multiplot grid pair plot
# Load iris dataset
iris = sns.load_dataset('iris')

# Basic pair plot
g1 = sns.pairplot(iris)
g1.fig.suptitle('Basic Pair Plot - Iris Dataset',
                fontsize=14, fontweight='bold', y=1.01)
plt.show()

# Pair plot with hue
g2 = sns.pairplot(iris, hue='species', palette='Set2')
g2.fig.suptitle('Pair Plot with Hue',
                fontsize=14, fontweight='bold', y=1.01)
plt.show()

# Pair plot with different diagonal plot
g3 = sns.pairplot(iris, hue='species', diag_kind='kde')
g3.fig.suptitle('Pair Plot with KDE on Diagonal',
                fontsize=14, fontweight='bold', y=1.01)
plt.show()
```
