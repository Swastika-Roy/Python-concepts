1. ARITHMETIC MEAN (Standard Average) (Σx)/n
2. GEOMETRIC MEAN (For Growth Rates)  (Πx)^(1/n)
3. HARMONIC MEAN (For Rates and Ratios) n/(Σ(1/x))
4. WEIGHTED MEAN (Values with Different Weights) (Σw*x)/Σw
5. TRIMMED MEAN (Excluding Outliers - 20% Trim)

6. MEDIAN (Middle value when sorted)

7. MODE (Most Frequent Value)
1. Unimodal: One mode
2. Bimodal: Two modes
3. Multimodal: Multiple modes
4. No mode: All values occur equally


**Use Mean when:**
- Data is normally distributed
- No extreme outliers
- Need to use all data points

**Use Median when:**
- Data has outliers
- Distribution is skewed
- Need a robust central tendency measure

**Use Mode when:**
- Data is categorical
- Need to find most frequent value
- Working with nominal data

```python
import numpy as np
import pandas as pd
data = [12,15,18,22,25,28,30,35,40,45]

#quartiles using numpy
q1=np.percentile(data,25)
q2=np.percentile(data,50)
q3=np.percentile(data,75)

print(f"Q1: (25th percentile) : {q1}")
print(f"Q2: (median,50th percentile) :{q2}")
print(f"Q3: (75th percentile) : {q3}")
print(f"IQR (Q3-Q1) : {q3-q1}")

```

```python
print("=" * 60)
print("COMPLETE STATISTICAL ANALYSIS")
print("=" * 60)

#RANGE
data_range = max(data) - min(data)
print(f"Range: {data_range}")


#variance and standard deviation
mean = np.mean(data)
pop_var = np.var(data)
sample_var = np.var(data,ddof=1)
pop_std = np.std(data)
sample_std=np.std(data,ddof=1)
print(f"Mean: {mean}")
print(f"Population Variance: {pop_var}")
print(f"Sample Variance: {sample_var}")
print(f"Population Standard Deviation: {pop_std}")
```
