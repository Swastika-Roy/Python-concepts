SELECT AND FILTER **DATA**

```python
df = pd.read_csv('/content/data.csv')
#selecting name and department columns
selected_cols = df[["name","department"]]
print(selected_cols)
```

```python
#selecting row with index 5 using .loc
row_lock = df.loc[5]
print(row_lock)
```

```python
#selecting first 3 rows and cols from 0 to 4 using .iloc
row_col_loc = df.iloc[0:3,0:5] #rows 0,1,2 and cols 0,1,2,3,4
print(row_col_loc)
```

```python
filtered_sal = df[df['salary']>70000]
print(filtered_sal)
```

```python
#filtering rows where depar is engineering and projects_completed > 10
filt_dept_pro=df[(df['department'] == 'Engineering')  & (df['projects_completed']>10)]
print(filt_dept_pro)
```

```python
#where city is new york
city=df[df['city']=='New York']
print(city)
```

```python
#remote employee is_remote = true
remote=df[df['is_remote']==True]
print(remote)
```

```python
#MEAN OF NUM VALUES GROUP BY DEPT, excluding 'employee_id'
mean_no_empid = df.drop(columns=['employee_id']).groupby('department').mean(numeric_only=True)
print(mean_no_empid)
```

```python
from os.path import samefile
#name and salary for employees in 'sales' dept using .loc
samefile=df.loc[df['department']=='Sales',['name','salary']]
print(samefile)
```

```python
#department is either marketing or hr
dept = df[df['department'].isin(['Marketing','HR'])]
print(dept)
```

```python
#name contains john (case-insensitive)
print( df[df['name'].str.contains('John',case=False,na=False)])
```

```python
#performance score > 9 and pro completed >=15
perf = df.loc[(df['performance_score']>9) & (df['projects_completed']>=15),['name','department','salary']]
print(perf)
```

```python
#except in 'sales' department
dept = df[df['department'] != 'Sales']
print(dept)
```

Missing value - NAN

```python
#MISSING VALUES
miss = df.isnull().sum()
print(miss)
```

```python
#handling missing values
df['salary'] = df['salary'].fillna(df['salary'].mean())
df['performance_score'] = df['performance_score'].fillna(df['performance_score'].mean())
#categorial cols like city ,fill with mode
df['city'] = df['city'].fillna(df['city'].mode()[0])
print("Misssing values : ")
print(df.isnull().sum())
```

```python
print(df)
```

GROUP AND AGGREGATION

>



```python
#MEAN OF NUM VALUES GROUP BY DEPT
# mean = df.groupby('department').mean(numeric_only=True)
# print(mean)
#MEAN OF NUM VALUES GROUP BY DEPT, excluding 'employee_id'
mean_no_empid = df.drop(columns=['employee_id']).groupby('department').mean(numeric_only=True)
print(mean_no_empid)
```

```python
#SUM OF NUM VALUES
sum=df.groupby('department').sum(numeric_only=True)
print(sum)
```

```python
#COUNT OF ENTRIES
count=df.groupby('department').size().reset_index(name='EmployeeCount')
print(count)
```

**VISUALISATION**

```python
import matplotlib.pyplot as plt
print("calc avg salary per dept: ")
avg_salary_dept = df.groupby('department')['salary'].mean()
print(avg_salary_dept)
```

```python
#create a bar chart of avg sal of dept
plt.figure(figsize=(10,6))
avg_salary_dept.plot.bar(color="skyblue")
# plt.xlabel('Department')
# plt.ylabel('Average Salary')
# plt.title('Average Salary per Department')
# plt.show()
plt.title("avg sal by dept")
plt.xlabel("department")
plt.ylabel("average salary")
plt.xticks(rotation=45,ha='right')#rotate x labels for better readability
# plt.yticks(rotation=45,ha='right')
plt.tight_layout()
plt.show()
```
