```python
#furniture or office supplies
df.loc[(df['category']=='Furniture') | (df['category']=='Office Supplies')]
```

```python
#consumer segment with express shipping
df.loc[(df['customer_segment']=='Consumer') & (df['ship_mode']=='Express')]
```

```python
#orders from east or west region with high value
df.loc[((df['region']=='East') | (df['region']=='West')) & (df['total_amount']>500)]
```

```python
#online electronics with discount
df.loc[(df['sales_channel']=='Online') & (df['category']=='Electronics') & (df['discount_percent']>0)]
```
