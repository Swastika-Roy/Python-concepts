```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline

#read data into a DataFrame
data = pd.read_csv('/content/Advertising.csv',index_col=0)
data.head()
```

```python
fig,axs = plt.subplots(1,3,sharey=True)
data.plot(kind="scatter",x='TV',y="Sales",ax=axs[0],figsize=(16,8))
data.plot(kind='scatter',x="Radio",y="Sales",ax=axs[1])
data.plot(kind='scatter',x="Newspaper",y="Sales",ax=axs[2])
```

```python
#select variables
X = data["TV"].values
Y = data["Sales"].values
n = len(X)
```

```python
#Calculate Mean (Manual)
x_mean = np.sum(X) / n
y_mean = np.sum(Y) / n
```

```python
print("Mean of X:", x_mean)
print("Mean of Y:", y_mean)
```

```python
# Calculate Slope (b1) Manually
numerator = np.sum((X - x_mean) * (Y - y_mean))
denominator = np.sum((X - x_mean) ** 2)
b1 = numerator / denominator
print("Slope (b1):", b1)
```

```python
# Calculate Intercept (b0)
b0 = y_mean - b1 * x_mean
print("Intercept (b0):", b0)
```

```python
# Prediction Formula
# y^​=b0​+b1​x

Y_pred = b0 + b1 * X
# Y_pred = b0 + b1 * 600
# print(Y_pred)
```

```python
# Error Calculation (Row-wise)
# Residuals
residuals = Y - Y_pred
# Absolute Error
absolute_error = np.abs(residuals)
# Squared Error
squared_error = residuals ** 2

# Error Metrics (Manual)
MAE = np.sum(absolute_error) / n
MSE = np.sum(squared_error) / n
RMSE = np.sqrt(MSE)

print("\n--- Model Parameters ---")
print("Intercept (b0):", b0)
print("Slope (b1):", b1)

print("\n--- Error Metrics ---")
print("MAE :", MAE)
print("MSE :", MSE)
print("RMSE:", RMSE)
```

```python
result = pd.DataFrame({
    "TV": X,
    "Actual Sales": Y,
    "Predicted Sales": Y_pred,
    "Residual": residuals,
    "Absolute Error": absolute_error,
    "Squared Error": squared_error
})

print(result.head(10))
```
