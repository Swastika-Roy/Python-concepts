**date filtering**

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])
```

```python
#orders in jan 2023
res = df[(df['order_date'] >= '2023-01-01') & (df['order_date'] < '2023-02-01')]
print(res)
```

```python
#orders after 2023 march
df.loc[df['order_date'] > '2023-03-01']
```

```python
#calculate shipping duration > 5 days
df['ship_date'] = pd.to_datetime(df['ship_date'])
df['order_date'] = pd.to_datetime(df['order_date'])
df['shipping_duration'] = (df['ship_date'] - df['order_date']).dt.days
df.loc[df['shipping_duration'] > 5]
```

**sorting and top/bottom values**

```python
#top 10 highest values orders
res = df.nlargest(10, 'total_amount')
print(res)
```

```python
#bottom 10 lowest value orders
df.nsmallest(10, 'total_amount')
```

```python
#sort by multiple cols
df.sort_values(['region', 'total_amount'], ascending=[False, True])
```

**aggregation and filtering**

```python
#total sales by category
df.groupby('category')['total_amount'].sum()
```

```python
#average order value by region
df.groupby('region')['total_amount'].mean().round(2)
```

```python
#count of orders by customer agent
df.groupby('customer_segment').value_counts()
```

```python
#total profit by product category
df.groupby('category')['profit'].sum().sort_values(ascending=False)
```

```python
#top 5 customers by total spending
df.groupby('customer_name')['total_amount'].sum().nlargest(5)
```

```python
#average discount by category
df.groupby('category')['discount_percent'].mean()
```

```python
#sales by payment method
df.groupby('payment_method')['total_amount'].sum()
```

**profit vs unprofit**

```python
#profitable vs unprofitable products
res = df.nlargest(10, 'profit')[["product_name","category","profit","total_amount"]]
print(res)
```

```python
#region wise category performance
df.groupby(['region', 'category'])['total_amount'].sum().unstack(fill_value=0)
```

```python
#corporate customers with high value orders
res = df[(df['customer_segment']=='Corporate') & (df['total_amount']>200)]
```

```python
#monthly sales trend
df['month'] = pd.to_datetime(df['order_date']).dt.month
df.groupby('month')['total_amount'].sum()
```

```python
#products with high quantity but low value
df.nlargest(10, ['quantity', 'total_amount'])
```

```python
#best performing sales chanels by region
df.groupby(['region', 'sales_channel'])['total_amount'].sum()
```

```python
#corporate customers high value orders
df.loc[(df['customer_segment']=='Corporate') & (df['total_amount']>500)]
```

```python
#profit margin analysis
df['profit_margin'] = (df['profit'] / df['total_amount'] * 100).round(2)
# df.head()
```

```python
#customer frequency analysis
customer_frequency = df['customer_name'].value_counts()
frequent_customers = customer_frequency[customer_frequency > 3]
print(frequent_customers)
```
